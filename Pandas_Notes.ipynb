# Pandas DataFrame Core Operations

Pandas is a Python library used for **working with structured and tabular data**.

It is very useful in:

- Data Science
- Machine Learning
- Data Analysis
- Artificial Intelligence
- Data Cleaning

The main data structure of Pandas is called a **DataFrame**.

A DataFrame stores data in the form of:

```
Rows + Columns
```

For example:

```
      Name    Age    Marks
0     Ali     20      80
1     Sara    21      90
2     Ahmed   19      75
```

We can imagine a DataFrame like an **Excel table**.

# 2. Creating a DataFrame

A DataFrame stores data in **rows and columns**.

We can create a DataFrame using a Python dictionary.

```python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Age": [20, 21, 19],
    "Marks": [80, 90, 75]
}

df = pd.DataFrame(data)

print(df)
```

Output:

```
    Name  Age  Marks
0    Ali   20     80
1   Sara   21     90
2  Ahmed   19     75
```

Here:

```
Name, Age, Marks → Columns

Ali, Sara, Ahmed → Data

0, 1, 2 → Index
```

A DataFrame is basically a **table inside Python**.

# 10. Filtering Data

Filtering means getting only the rows that match a condition.

Suppose:

```python
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Marks": [45, 90, 75, 40]
}

df = pd.DataFrame(data)
```

To get students with marks greater than or equal to `50`:

```python
passed = df[df["Marks"] >= 50]

print(passed)
```

Output:

```
    Name  Marks
1   Sara     90
2  Ahmed     75
```

To get failed students:

```python
failed = df[df["Marks"] < 50]

print(failed)
```

Output:

```
     Name  Marks
0     Ali     45
3  Ayesha     40
```

Filtering is one of the most important operations in data analysis.

# 14. Handling Missing Values

Real datasets often contain missing data.

Example:

```python
data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Marks": [80, None, 75]
}

df = pd.DataFrame(data)

print(df)
```

Output:

```
    Name  Marks
0    Ali   80.0
1   Sara    NaN
2  Ahmed   75.0
```

`NaN` means a value is missing.

---

## Check Missing Values

```python
print(df.isnull())
```

To count missing values:

```python
print(df.isnull().sum())
```

---

## Remove Missing Values

```python
df = df.dropna()
```

`dropna()` removes rows containing missing values.

---

## Fill Missing Values

Instead of deleting them:

```python
df["Marks"] = df["Marks"].fillna(0)
```

We can also fill missing marks using average marks:

```python
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())
```

Missing-value handling is an important part of **data cleaning**.

# 15. Handling Duplicate Data

Sometimes the same record appears more than once.

Example:

```python
data = {
    "Name": ["Ali", "Sara", "Ali"],
    "Marks": [80, 90, 80]
}

df = pd.DataFrame(data)
```

Check duplicates:

```
False
False
True
```

The last row is a duplicate.

To count duplicates:

```python
print(df.duplicated().sum())
```

To remove duplicates:

```python
df = df.drop_duplicates()

print(df)
```

Duplicate removal is also part of **data cleaning**.

# 16. Sorting Data

`sort_values()` arranges data in a particular order.

Example:

```python
data = {
    "Name": ["Ali", "Sara", "Ahmed"],
    "Marks": [70, 95, 80]
}

df = pd.DataFrame(data)
```

Sort from lowest to highest:

```python
print(df.sort_values("Marks"))
```

Output:

```
    Name  Marks
0    Ali     70
2  Ahmed     80
1   Sara     95
```

To sort from highest to lowest:

```python
print(df.sort_values("Marks", ascending=False))
```

Output:

```
    Name  Marks
1   Sara     95
2  Ahmed     80
0    Ali     70
```

So:

```
ascending=True  → Smallest to Largest

ascending=False → Largest to Smallest
```

## groupby()

`groupby()` allows us to analyze data according to categories.

Example:

```python
data = {
    "Name": ["Ali", "Sara", "Ahmed", "Ayesha"],
    "Course": ["AI", "Web", "AI", "Web"],
    "Marks": [80, 90, 70, 85]
}

df = pd.DataFrame(data)
```

Find average marks for each course:

```python
result = df.groupby("Course")["Marks"].mean()

print(result)
```

Output:

```
Course
AI     75.0
Web    87.5
```

It works like:

```
AI Students
    ↓
80, 70
    ↓
Average = 75

Web Students
    ↓
90, 85
    ↓
Average = 87.5
```

`groupby()` is very important in Data Analysis.

# 18. Combining DataFrames

Sometimes our data exists in multiple DataFrames.

Pandas allows us to combine them.

## concat()

Suppose:

```python
df1 = pd.DataFrame({
    "Name": ["Ali", "Sara"],
    "Marks": [80, 90]
})

df2 = pd.DataFrame({
    "Name": ["Ahmed", "Ayesha"],
    "Marks": [75, 85]
})
```

Combine them:

```python
final_df = pd.concat([df1, df2], ignore_index=True)

print(final_df)
```

Output:

```
     Name  Marks
0     Ali     80
1    Sara     90
2   Ahmed     75
3  Ayesha     85
```

`concat()` is useful when we want to **join similar datasets vertically**.

---

## merge()

`merge()` combines DataFrames using a common column.

Example:

```python
students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Ali", "Sara", "Ahmed"]
})

marks = pd.DataFrame({
    "ID": [1, 2, 3],
    "Marks": [80, 90, 75]
})
```

Merge using `ID`:

```python
final_df = pd.merge(students, marks, on="ID")

print(final_df)
```

Output:

```
   ID   Name  Marks
0   1    Ali     80
1   2   Sara     90
2   3  Ahmed     75
```

The common column is:

```
ID
```

So:

```
Students Data
      +
Marks Data
      ↓
Common ID
      ↓
Final DataFrame
```

# Class Practice: Student Performance Analyzer

Now we will use different Pandas concepts in one small program.

## Problem

We have information about different students.

We want to find:

- Average marks
- Highest marks
- Lowest marks
- Passed students
- Failed students
- Course-wise average
- Add a Result column

## Code

```python
import pandas as pd

data = {
    "Name": [
        "Ali",
        "Sara",
        "Ahmed",
        "Ayesha",
        "Zain"
    ],

    "Course": [
        "AI",
        "Web",
        "AI",
        "Web",
        "AI"
    ],

    "Marks": [
        45,
        90,
        75,
        85,
        40
    ]
}

df = pd.DataFrame(data)

print("Student Data:")
print(df)

print("\nAverage Marks:")
print(df["Marks"].mean())

print("\nHighest Marks:")
print(df["Marks"].max())

print("\nLowest Marks:")
print(df["Marks"].min())

passed = df[df["Marks"] >= 50]

failed = df[df["Marks"] < 50]

print("\nPassed Students:")
print(passed)

print("\nFailed Students:")
print(failed)

df["Result"] = df["Marks"].apply(
    lambda marks: "Pass" if marks >= 50 else "Fail"
)

print("\nFinal Data:")
print(df)

print("\nCourse Wise Average:")
print(df.groupby("Course")["Marks"].mean())
```
