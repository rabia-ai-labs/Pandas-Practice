## Task 1

Create this student DataFrame:

```
Name     Marks
Ali       70
Sara      85
Ahmed     60
Ayesha    95
Zain      45
```

Find:

- Average marks
- Maximum marks
- Minimum marks
- Total students

---

````python
#TASK 1

import pandas as pd

data = {
    "Name": ['Ali','Sarah','Ahmed','Ayesha','Zain'],
    "Marks": [70,85,60,95,45]
}

df = pd.DataFrame(data)
print(df)

print(f"\nAverage Marks: {df['Marks'].mean()}")
print(f"Maximum Marks: {df['Marks'].max()}")
print(f"Minimum Marks: {df['Marks'].min()}")
print(f"Total Students: {df['Name'].count()}")
````

## Task 2

Using the same DataFrame:

Display only students whose marks are greater than or equal to `50`.

Then display students whose marks are less than `50`.

---


````python
# TASK 2

print(df[df['Marks'] >= 50])
print("\n")
print(df[df['Marks'] < 50])
````

## Task 4

Create a DataFrame containing:

```
Name
Course
Marks
```

Use these courses:

```
AI
Web
AI
Web
AI
```

Find the **average marks of each course** using `groupby()`.

---


````python
# Task 4
data = {
    "Name": ['Ali','Sarah','Rabia','Sana','Bilal'],
    "Course": ['AI','Web','AI','Web','AI'],
    "Marks": [60,80,94,91,73]
}

df = pd.DataFrame(data)
print(df)

avg_marks = df.groupby('Course')['Marks'].mean()
print(f"\nAverage Marks of each course: {avg_marks}")
````

## Task 5

Create a DataFrame containing some missing marks.

For example:

```
Ali       80
Sara      NaN
Ahmed     70
Ayesha    NaN
Zain      90
```

Find:

- Number of missing values
- Fill missing values with average marks
- Display the cleaned DataFrame

---


````python
import pandas as pd

#TASK 5

data = {
    "Name": ['Ali', 'Sara', 'Ahmed', 'Ayesha', 'Zain'],
    "Marks": [80, None, 70, None, 90]
}

df = pd.DataFrame(data)
print("Original DataFrame:")
print(df)

# Count missing values
print("\nMissing Values:")
print(df['Marks'].isnull().sum())

# Calculate average marks
avg_marks = df['Marks'].mean()
print("\nAverage Marks:", avg_marks)

# Fill missing values with average
df['Marks'] = df['Marks'].fillna(avg_marks)

print("\nCleaned DataFrame:")
print(df)
````
